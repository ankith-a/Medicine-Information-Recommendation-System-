```python
def recommend_medicine(disease=None, treatment=None, tablet=None, top_n=5):
    
    result = combination_frequency.copy()

    # Disease filter
    if disease is not None:
        result = result[
            result['Disease'].str.lower() == disease.strip().lower()
        ]

    # Treatment filter
    if treatment is not None:
        result = result[
            result['Treatment'].str.lower() == treatment.strip().lower()
        ]

    # Tablet filter
    if tablet is not None:
        result = result[
            result['Tablet Name'].str.lower() == tablet.strip().lower()
        ]

    # No results
    if result.empty:
        print("No matching medicine combination found.")
        return None

    # Rank by frequency
    result = result.sort_values(
        'Frequency',
        ascending=False
    ).head(top_n)

    return result.reset_index(drop=True)
```

```python
def recommend_medicine_details(disease=None, treatment=None, tablet=None, top_n=5):
    
    # Get ranked combinations
    recommendations = recommend_medicine(
        disease=disease,
        treatment=treatment,
        tablet=tablet,
        top_n=top_n
    )

    if recommendations is None:
        return None

    # Get complete details from original cleaned dataset
    details = medicine_data.merge(
        recommendations[
            ['Disease', 'Treatment', 'Tablet Name', 'Frequency']
        ],
        on=['Disease', 'Treatment', 'Tablet Name'],
        how='inner'
    )

    # Remove exact duplicate detail records
    details = details.drop_duplicates()

    # Arrange columns
    details = details[
        [
            'Disease',
            'Treatment',
            'Tablet Name',
            'Dosage',
            'Time Taken',
            'Effect Time Duration',
            'Which Age',
            'Company',
            'Purity',
            'Side Effects',
            'Price',
            'Frequency'
        ]
    ]

    return details
```

```python
def flexible_medicine_search(disease=None, treatment=None, tablet=None, top_n=5):
    
    result = medicine_summary.copy()

    # Disease search
    if disease is not None:
        result = result[
            result['Disease'].str.contains(
                disease.strip(),
                case=False,
                na=False
            )
        ]

    # Treatment search
    if treatment is not None:
        result = result[
            result['Treatment'].str.contains(
                treatment.strip(),
                case=False,
                na=False
            )
        ]

    # Tablet search
    if tablet is not None:
        result = result[
            result['Tablet Name'].str.contains(
                tablet.strip(),
                case=False,
                na=False
            )
        ]

    # No results
    if result.empty:
        print("No matching medicine information found.")
        return None

    # Rank results
    result = (
        result
        .sort_values('Frequency', ascending=False)
        .head(top_n)
        .reset_index(drop=True)
    )

    return result
```

```python
def validated_medicine_search(disease=None, treatment=None, tablet=None, top_n=5):

    # Count how many inputs were provided
    input_count = sum(
        value is not None and value.strip() != ""
        for value in [disease, treatment, tablet]
    )

    # Validation: no input
    if input_count == 0:
        print("Please enter at least one input: Disease, Treatment, or Tablet.")
        return None

    # Validation: more than two inputs
    if input_count > 2:
        print("Please enter only one or two inputs.")
        return None

    # Call flexible search
    return flexible_medicine_search(
        disease=disease,
        treatment=treatment,
        tablet=tablet,
        top_n=top_n
    )
```

```python
def predict_medicine(disease=None, treatment=None, tablet=None, top_n=5):

    # Check that at least one input is provided
    inputs = [
        value.strip()
        for value in [disease, treatment, tablet]
        if value is not None and value.strip() != ""
    ]

    if len(inputs) == 0:
        print("Please provide at least one input.")
        return None

    if len(inputs) > 2:
        print("Please provide only one or two inputs.")
        return None

    # Combine user inputs
    user_query = " ".join(inputs)

    # Convert user query into TF-IDF vector
    query_vector = tfidf.transform([user_query])

    # Calculate similarity with all medicine combinations
    similarity_scores = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    # Get indexes of highest similarity scores
    top_indices = similarity_scores.argsort()[::-1][:top_n]

    # Create result
    results = model_data.iloc[top_indices].copy()

    # Add similarity score
    results['Similarity Score'] = similarity_scores[top_indices]

    # Reset index
    results = results.reset_index(drop=True)

    return results
```

```python
def evaluate_recommendation(query, expected_disease=None,
                            expected_treatment=None,
                            expected_tablet=None,
                            top_n=5):

    results = predict_medicine(
        disease=query.get('disease'),
        treatment=query.get('treatment'),
        tablet=query.get('tablet'),
        top_n=top_n
    )

    if results is None:
        return False

    # Check whether expected combination appears
    match = False

    for _, row in results.iterrows():

        disease_match = (
            expected_disease is None or
            row['Disease'].lower() == expected_disease.lower()
        )

        treatment_match = (
            expected_treatment is None or
            row['Treatment'].lower() == expected_treatment.lower()
        )

        tablet_match = (
            expected_tablet is None or
            row['Tablet Name'].lower() == expected_tablet.lower()
        )

        if disease_match and treatment_match and tablet_match:
            match = True
            break

    return match
```

```python
def evaluate_model(test_data, input_type, top_n=5):
    
    correct = 0
    total = len(test_data)

    for _, row in test_data.iterrows():

        # Disease only
        if input_type == 'Disease':
            result = predict_medicine(
                disease=row['Disease'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    result['Disease'].str.lower()
                    == row['Disease'].lower()
                ).any()

        # Treatment only
        elif input_type == 'Treatment':
            result = predict_medicine(
                treatment=row['Treatment'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    result['Treatment'].str.lower()
                    == row['Treatment'].lower()
                ).any()

        # Tablet only
        elif input_type == 'Tablet':
            result = predict_medicine(
                tablet=row['Tablet Name'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    result['Tablet Name'].str.lower()
                    == row['Tablet Name'].lower()
                ).any()

        # Disease + Treatment
        elif input_type == 'Disease + Treatment':
            result = predict_medicine(
                disease=row['Disease'],
                treatment=row['Treatment'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    (result['Disease'].str.lower() == row['Disease'].lower()) &
                    (result['Treatment'].str.lower() == row['Treatment'].lower())
                ).any()

        # Disease + Tablet
        elif input_type == 'Disease + Tablet':
            result = predict_medicine(
                disease=row['Disease'],
                tablet=row['Tablet Name'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    (result['Disease'].str.lower() == row['Disease'].lower()) &
                    (result['Tablet Name'].str.lower() == row['Tablet Name'].lower())
                ).any()

        # Tablet + Treatment
        elif input_type == 'Tablet + Treatment':
            result = predict_medicine(
                tablet=row['Tablet Name'],
                treatment=row['Treatment'],
                top_n=top_n
            )

            if result is not None:
                match = (
                    (result['Tablet Name'].str.lower() == row['Tablet Name'].lower()) &
                    (result['Treatment'].str.lower() == row['Treatment'].lower())
                ).any()

        else:
            raise ValueError("Invalid input type.")

        if match:
            correct += 1

    return correct / total
```

```python
input_types = [
    'Disease',
    'Treatment',
    'Tablet',
    'Disease + Treatment',
    'Disease + Tablet',
    'Tablet + Treatment'
]

results = []

for input_type in input_types:
    
    accuracy = evaluate_model(
        test_data,
        input_type,
        top_n=5
    )
    
    results.append({
        'Input Type': input_type,
        'Top-5 Accuracy': round(accuracy * 100, 2)
    })

evaluation_results = pd.DataFrame(results)

evaluation_results
```

```python
def predict_medicine_knn(
    disease=None,
    treatment=None,
    tablet=None,
    top_n=5
):

    # Collect valid inputs
    inputs = [
        value.strip()
        for value in [disease, treatment, tablet]
        if value is not None and value.strip() != ""
    ]

    # Validate input
    if len(inputs) == 0:
        print("Please provide at least one input.")
        return None

    if len(inputs) > 2:
        print("Please provide only one or two inputs.")
        return None

    # Combine inputs
    user_query = " ".join(inputs)

    # Convert query into TF-IDF
    query_vector = tfidf.transform([user_query])

    # Find nearest medicine combinations
    distances, indices = knn_model.kneighbors(
        query_vector,
        n_neighbors=min(top_n, len(model_data))
    )

    # Convert cosine distance to similarity
    similarity_scores = 1 - distances[0]

    # Get results
    results = model_data.iloc[indices[0]].copy()

    # Add similarity score
    results['Similarity Score'] = similarity_scores

    # Reset index
    results = results.reset_index(drop=True)

    return results
```

```python
def evaluate_model_predictions(test_data, model_type, top_n=5):

    top1_correct = 0
    top5_correct = 0
    similarity_scores = []

    for _, row in test_data.iterrows():

        # Model 1: TF-IDF + Cosine Similarity
        if model_type == 'Cosine':

            result = predict_medicine(
                disease=row['Disease'],
                treatment=row['Treatment'],
                top_n=top_n
            )

        # Model 2: TF-IDF + KNN
        elif model_type == 'KNN':

            result = predict_medicine_knn(
                disease=row['Disease'],
                treatment=row['Treatment'],
                top_n=top_n
            )

        else:
            raise ValueError("Unknown model type.")

        if result is None or len(result) == 0:
            continue

        # Check Disease + Treatment match
        matches = (
            (result['Disease'].str.lower() == row['Disease'].lower()) &
            (result['Treatment'].str.lower() == row['Treatment'].lower())
        )

        # Top-1
        if matches.iloc[0]:
            top1_correct += 1

        # Top-5
        if matches.any():
            top5_correct += 1

        # Store similarity scores
        similarity_scores.extend(
            result['Similarity Score'].tolist()
        )

    total = len(test_data)

    if total == 0:
        return {
            'Top-1 Accuracy (%)': 0,
            'Top-5 Accuracy (%)': 0,
            'Average Similarity': 0
        }

    return {
        'Top-1 Accuracy (%)': round(
            (top1_correct / total) * 100, 2
        ),
        'Top-5 Accuracy (%)': round(
            (top5_correct / total) * 100, 2
        ),
        'Average Similarity': round(
            sum(similarity_scores) / len(similarity_scores), 4
        )
        if similarity_scores else 0
    }
```

```python
# ============================================
# RANDOM FOREST MODEL
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report


# 1. Create model dataset
rf_data = medicine_data[
    ['Disease', 'Tablet Name', 'Treatment']
].drop_duplicates()


# 2. Remove Treatment classes with fewer than 2 samples
treatment_counts = rf_data['Treatment'].value_counts()

valid_treatments = treatment_counts[
    treatment_counts >= 2
].index

rf_data = rf_data[
    rf_data['Treatment'].isin(valid_treatments)
].copy()


# 3. Define input and target
X = rf_data[['Disease', 'Tablet Name']]
y = rf_data['Treatment']


# 4. Display dataset information
print("Model dataset:", rf_data.shape)
print("Number of Treatment classes:", y.nunique())


# 5. Train-test split
n_classes = y.nunique()

test_samples = max(
    n_classes,
    int(len(X) * 0.25)
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=test_samples,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))


# 6. One-hot encoding
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(handle_unknown='ignore'),
            ['Disease', 'Tablet Name']
        )
    ]
)


# 7. Create Random Forest
rf_model = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        (
            'classifier',
            RandomForestClassifier(
                n_estimators=200,
                random_state=42,
                class_weight='balanced'
            )
        )
    ]
)


# 8. Train model
rf_model.fit(X_train, y_train)

print("\nRandom Forest model trained successfully!")
```

```python
# Make predictions
y_pred = rf_model.predict(X_test)

# Accuracy
rf_accuracy = accuracy_score(y_test, y_pred)

print(
    "Random Forest Accuracy:",
    round(rf_accuracy * 100, 2),
    "%"
)

# Detailed report
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
def predict_treatment_rf_with_probability(disease, tablet):
    
    new_data = pd.DataFrame({
        'Disease': [disease],
        'Tablet Name': [tablet]
    })
    
    # Predicted treatment
    prediction = rf_model.predict(new_data)[0]
    
    # Prediction probabilities
    probabilities = rf_model.predict_proba(new_data)[0]
    
    # Treatment class names
    classes = rf_model.classes_
    
    # Create probability table
    probability_data = pd.DataFrame({
        'Treatment': classes,
        'Probability': probabilities
    })
    
    probability_data = probability_data.sort_values(
        'Probability',
        ascending=False
    ).reset_index(drop=True)
    
    return prediction, probability_data
```
